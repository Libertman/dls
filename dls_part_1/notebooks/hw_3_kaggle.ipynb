```python
# Для вашего удобства списки с именами разных колонок

# Числовые признаки
num_cols = [
    'ClientPeriod',
    'MonthlySpending',
    'TotalSpent'
]

# Категориальные признаки
cat_cols = [
    'Sex',
    'IsSeniorCitizen',
    'HasPartner',
    'HasChild',
    'HasPhoneService',
    'HasMultiplePhoneNumbers',
    'HasInternetService',
    'HasOnlineSecurityService',
    'HasOnlineBackup',
    'HasDeviceProtection',
    'HasTechSupportAccess',
    'HasOnlineTV',
    'HasMovieSubscription',
    'HasContractPhone',
    'IsBillingPaperless',
    'PaymentMethod'
]

feature_cols = num_cols + cat_cols
target_col = 'Churn'
```

```python
def postprocessing_drop(df):
    '''удаление незначимых признаков после обработки данных'''

    return df.drop(columns=['Sex', 'HasPhoneService', 'HasMultiplePhoneNumbers', 'HasOnlineTV', 'HasMoviesSubscription', 'HasInternetService'], errors='ignore')
```

```python
num_cols.remove('TotalSpent') if 'TotalSpent' in num_cols else None

ord_cols = ['HasContractPhone']

for col in ['Sex', 'HasPhoneService', 'HasMultiplePhoneNumbers', 'HasOnlineTV', 'HasMoviesSubscription', 'HasInternetService']:
    cat_cols.remove(col) if col in cat_cols else None
cat_cols.remove('HasContractPhone') if col in cat_cols else None
```

```python
ord_cols = ['HasContractPhone']
```

### **Реализуем пайплайн обработки данных для каждого типа колонки**

```python
num_pipeline = Pipeline(
    [
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]
)

ord_pipeline = Pipeline(
    [
        ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
        ('ordinal', OrdinalEncoder(categories=[['Month-to-month', 'One year', 'Two year']]))
    ]
)

cat_pipeline = Pipeline(
    [
        ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
        ('encoder', OneHotEncoder(drop='if_binary', handle_unknown='ignore', sparse_output=False)),
    ]
)

total_spent_pipeline = Pipeline(
    [
        ('imputer', SimpleImputer(strategy='median')),
        ('box-cox', PowerTransformer(method='box-cox')),
        ('scaler', StandardScaler())
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('total_spent',total_spent_pipeline, ['TotalSpent']),
        ('cat', cat_pipeline, cat_cols),
        ('num', num_pipeline, num_cols),
        ('ord', ord_pipeline, ord_cols)
    ]
)
```

```python
def pipeline_model(decision_model):
    pipeline = Pipeline([
        ('postprocessing_drop', FunctionTransformer(postprocessing_drop, validate=False)),
        ('preprocessor', preprocessor),
        ('classifier', decision_model)
    ])
    pipeline = pipeline.set_output(transform='pandas')
    return pipeline
```
